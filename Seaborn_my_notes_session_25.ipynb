```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

```python


tips = sns.load_dataset('tips')
tips.head(3)
```

```python
# scatter plot -> axes level function
sns.scatterplot(data = tips , x = 'total_bill' , y = 'tip')
```

```python
# hue , style , size
# hue - use for grouping the values on the basis of a column ,  and then represent it in different colors(male , female)
# style - different stules of dots will appear according to the given column
# size - dot size represnts the amount pr size

sns.scatterplot(
    data = tips ,
    x = 'total_bill' ,
    y = 'tip' ,
    hue = 'sex' ,
    style = 'time',
    size='size'
    )
```

```python
# relplot -> figure level -> square shape

sns.relplot(data = tips , x = 'total_bill' , y = 'tip' , kind = 'scatter')
```

```python
sns.relplot(
    data = tips ,
    x = 'total_bill' ,
    y = 'tip',
    hue = 'sex',
    style = 'time',
    size = 'size'
    )
```

###NOTE - Most of the time , use the figure level function like relplot

### Axes level function comes under figures , figure is on the top

```python
# line plot

gap = px.data.gapminder()
temp_df = gap[gap['country'] == 'India']
temp_df
```

```python
# axes level function

sns.lineplot(data = temp_df , x = 'year' , y = 'lifeExp')
```

```python
# using relplot - figure level function

sns.relplot(data = temp_df , x = 'year' , y = 'lifeExp' , kind = 'line')
```

```python
temp_df = gap[gap['country'].isin(['India' , 'Brazil' , 'Germany'])]
temp_df
```

### One more difference between axes and figure is that , figure put the lagend outside the graph , and axes put the lagen inside the graph

```python
# hue -> style
# figure level
sns.relplot(
    kind = 'line' ,
    data  =temp_df ,
    x = 'year' ,
    y = 'lifeExp' ,
    hue = 'country',
    style = 'continent',
    size = 'continent'          # thickness of the line
    )
```

```python
# axes level

sns.lineplot(data=temp_df, x='year', y='lifeExp', hue='country')

# see , lagend is inside the graph
```

```python
# facet plot -> figure level function -> work with relplot
# it will not work with scatterplot and lineplot

sns.relplot(
    data = tips,
    x = 'total_bill',
    y = 'tip',
    kind = 'scatter',
    row = 'day',
    col = 'smoker',

    hue = 'sex',
    style = 'day',
    size = 'size'
    )


# col - it mean it makes columns by seeing the smoker column (2 column , for yes and no)
# row - it makes rows according to day (4 rows , thur fri sat sun)

# we can also use hue and style and size
```

```python
# col wrap

sns.relplot(
    data = gap ,
    x = 'lifeExp' ,
    y = 'gdpPercap' ,
    kind = 'scatter',
    col = 'year',
    col_wrap=3,
    )


# col_wrap = 4 - means make only 4 columns , if columns increased from 4 , then wrap it below
```

### 2. Distribution Plots

- used for univariate analysis
- used to find out the distribution
- Range of the observation
- Central Tendency
- is the data bimodal?
- Are there outliers?

Plots under distribution plot

- histplot
- kdeplot
- rugplot

```python
# figure level -> displot
# axes level -> histplot -> kdeplot -> rugplot
```

```python
# plotting univariate histogram

# by histplot

sns.histplot(data = tips , x = 'total_bill')
```

```python
# by distplot - it is figure level

sns.displot(data=tips , x = 'total_bill' , kind ='hist')
```

```python
# bins parameter

sns.displot(
    data=tips ,
    x = 'total_bill' ,
    kind ='hist',
    bins =40

    )
```

```python
# It’s also possible to visualize the distribution of a categorical variable using the logic of a histogram.
# Discrete bins are automatically set for categorical variables

# it is like a countplot

sns.displot(
    data = tips ,
    x = 'day' ,
    kind = 'hist',

    )
```

```python
# hue parameter

sns.displot(
    data=tips ,
    x = 'total_bill' ,
    kind ='hist',
    bins =40,
    hue='sex'

    )
```

```python
# element -> step

sns.displot(
    data=tips ,
    x = 'total_bill' ,
    kind ='hist',
    bins =40,
    hue = 'sex',
    element='step'

    )
```

```python
titanic = sns.load_dataset('titanic')
titanic
```

```python
sns.displot(
    data=titanic,
    x='age',
    kind='hist',
    element='step',
    hue='sex')
```

```python
# faceting using col and row -> not work on histplot function

sns.displot(
    data = titanic ,
    x = 'age' ,
    kind = 'hist',
    element = 'step',
    hue = 'sex',
    col = 'sex'
    )
```

```python
# kdeplot - axes level
# Rather than using discrete bins, a KDE plot smooths the observations with a Gaussian kernel, producing a continuous density estimate

sns.kdeplot(
    data = tips,
    x='total_bill'
)
```

```python
# figure level

sns.displot(
    kind = 'kde',
    data = tips,
    x = 'total_bill'
)
```

```python
# hue -> fill

sns.displot(
    kind = 'kde',
    data = tips,
    x = 'total_bill',
    hue  ='sex',
    fill = True
)
```

```python
# Rugplot

# Plot marginal distributions by drawing ticks along the x and y axes.

# This function is intended to complement other plots by showing the location of individual observations in an unobtrusive way.

sns.kdeplot(data = tips,x = 'total_bill')
sns.rugplot(data = tips,x = 'total_bill')
```
