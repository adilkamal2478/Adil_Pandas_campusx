### Boxplot

A boxplot is a standardized way of displaying the distribution of data based on a five number summary (“minimum”, first quartile [Q1], median, third quartile [Q3] and “maximum”). It can tell you about your outliers and what their values are. Boxplots can also tell you if your data is symmetrical, how tightly your data is grouped and if and how your data is skewed.

Categorical Estimate Plot -> for central tendency


*   Barplot
*   Pointplot
*   Countplot




```python
# barplot
# A bar plot shows the mean of a numeric variable for each category, with optional error bars.


# axes level function
import numpy as np
sns.barplot(
    data = tips,
    x = 'sex',
    y = 'total_bill',
    hue = 'smoker',
    # estimator=np.min ,
    estimator = np.median
)

# estimator - any aggregate function name
```

```python
# figure level functon

sns.catplot(
    kind = 'bar' ,
    data = tips,
    x = 'sex',
    y = 'total_bill',
    hue = 'smoker',
    # estimator=np.min ,
    estimator = np.min
)
```

```python
sns.catplot(
    kind = 'bar' ,
    data = tips,
    x = 'sex',
    y = 'total_bill',
    hue = 'smoker',
    # estimator=np.min ,
    estimator = np.median ,
    ci = None
)

```

```python
# point plot
# it is brother of bar plot , where it takes the top part of all grapgh and make difference line between then
# it is not a simple line grapgh , it wroks on aggreagte funct
# simply understand ,  joint all top of bar plot , u will get this

sns.pointplot(
    data = tips,
    x = 'day',
    y = 'total_bill',
    errorbar = ('ci' , 0),
    estimator = np.min,
    hue = 'sex'
)
```

### **Point plot** - When there are multiple observations in each category, it also uses bootstrapping to compute a confidence interval around the estimate, which is plotted using error bars

```python
# countplot

sns.countplot(
    data = tips,
    x = 'sex',
    hue = 'day'
)

# returns the number of counts
```

**Count plot** - A special case for the bar plot is when you want to show the number of observations in each category rather than computing a statistic for a second variable. This is similar to a histogram over a categorical, rather than quantitative, variable

```python
# faceting using catplot

sns.catplot(
    data = tips ,
    x = 'sex',
    y = 'total_bill',
    kind = 'box' ,
    col = 'smoker',
    row = 'time',
    hue = 'sex'
)
```

### Regression Plots

- regplot
- lmplot

In the simplest invocation, both functions draw a scatterplot of two variables, x and y, and then fit the regression model y ~ x and plot the resulting regression line and a 95% confidence interval for that regression.

```python
# normal scatter plot
sns.scatterplot(
    data = tips,
    x = 'total_bill',
    y = 'tip'
)
```

```python
# regplot - add a extra regression line in the scatter plot
# axes level
# hue parameter is not available

sns.regplot(
    data = tips,
    x = 'total_bill',
    y = 'tip',
    # hue = 'sex'
)
```

```python
# lmplot - Figure-level function
# supports hue

sns.lmplot(
    data = tips,
    x = 'total_bill',
    y = 'tip',
    hue = 'sex'
)
```

```python
# residplot

# sns.residplot() is used to plot residuals of a regression — that is,
# the difference between observed values and the fitted regression line.

sns.residplot(
    data = tips,
    y = 'tip',
    x = 'total_bill'
)
```

### Plotting Pairwise Relationship (PairGrid Vs Pairplot)

```python
# pairplot
# pairplot = grid of scatterplots + histograms for all variable pairs.
# it automatically takes only numeric columns

sns.pairplot(iris)
```

```python
#  hue

sns.pairplot(iris , hue = 'species')
```

```python
# diag_kind - change the type of diagonal
# sns.pairplot(iris )
sns.pairplot(iris ,  diag_kind  = 'kde')

# diag_kind - it default value is hist , u can give kde also
```


`PairGrid` is the underlying class that powers `pairplot`. It gives you more manual control over how the grid of plots is built.

```python
# Pair grid - father of pairplot
# we can more customize the pairplot

g = sns.PairGrid(data = iris , hue='species')
g.map(sns.scatterplot)
```

```python
# map_diag -> map_offdiag

# diagonal graphs are boxplot , and remaining

g = sns.PairGrid(data = iris , hue='species')
g.map_diag(sns.boxplot)
g.map_offdiag(sns.kdeplot)
```

```python
# map_diag -> map_upper -> map_lower

# diagonal are histogram
# upper part of diagonal are scatterplot
# lower part is kde

g = sns.PairGrid(data = iris , hue='species')
g.map_upper(sns.scatterplot)
g.map_lower(sns.kdeplot, fill=True)
g.map_diag(sns.histplot)
```

```python
# vars - only specific columns

iris
g = sns.PairGrid(iris , hue = 'species', vars = ['sepal_width' , 'petal_width'])
g.map_upper(sns.kdeplot)
g.map_lower(sns.scatterplot)
g.map_diag(sns.histplot)
```

### JointGrid Vs Jointplot

`jointplot` = scatterplot + marginal distributions.

```python
# scatterplot

sns.jointplot(
    data = tips,
    x = 'total_bill',
    y = 'tip',
    kind = 'hist'
)

# it merges the x axis and y axis
```

```python
# kind can take - scatter(default) , reg , hex , kde

sns.jointplot(
    data = tips,
    x = 'total_bill',
    y = 'tip',
    kind = 'kde',
)
```

`JointGrid` = customizable layout for one joint plot + two marginals.

```python
g = sns.JointGrid(data=tips,x='total_bill',y='tip')
g.plot(sns.kdeplot,sns.violinplot)
```


* g.plot_joint(sns.scatterplot) → central plot (relationship between x and y).

* g.plot_marginals(sns.histplot) → side plots (distribution of x and y separately).

* g.plot_joint(sns.regplot) → regression line in the center.



```python
g = sns.JointGrid(data=tips,x='total_bill',y='tip')

g.plot_joint(sns.scatterplot)
g.plot_marginals(sns.histplot)
```
