# All figure function

* `relplot` → for relational plots (scatter, line).

* `displot` → for distributions (histograms, KDEs, rug plots).

* `catplot` → for categorical plots (strip, swarm, bar, box, etc.).



```python
import seaborn as sns
```

```python
# import datasets
tips = sns.load_dataset('tips')
iris = sns.load_dataset('iris')
```

```python
sns.scatterplot(data = tips , x = 'total_bill' , y = 'tip')
```

```python
# strip plot
# A strip plot is one of Seaborn’s categorical scatter plots
# axes level function

sns.stripplot(
    data = tips,
    x = 'day',
    y = 'total_bill',
)
```

```python
# jitter
# it remove unwanted jitter produced by seaborn in the graph

sns.stripplot(
    data = tips,
    x = 'day',
    y = 'total_bill',
    jitter = False
)
```

```python
# using catplot
# figure level function

sns.catplot(
    data = tips,
    x = 'day',
    y = 'total_bill',
    kind = 'strip'
)
```

```python
# changing jitter

sns.catplot(
    data = tips,
    x = 'day',
    y = 'total_bill',
    kind = 'strip',
    jitter = 0.2
)

# u can control the jitter as per ur requirement
```

```python
# hue

sns.catplot(
    data = tips,
    x = 'day',
    y = 'total_bill',
    kind = 'strip',
    hue = 'sex'
)
```

```python
# swarmplot - exactly same as strip plot
# Swarm plots automatically adjust the position of points so they don’t overlap.

# axes level
sns.swarmplot(data = tips , x = 'day' , y = 'total_bill')

# we can use this but for smaller data
```

```python
# figure level

sns.catplot( kind = 'swarm', data = tips , x = 'day' , y = 'total_bill')
```

```python
# hue

sns.catplot( kind = 'swarm', data = tips , x = 'day' , y = 'total_bill' , hue = 'sex')
```

### Boxplot

A boxplot is a standardized way of displaying the distribution of data based on a five number summary (“minimum”, first quartile [Q1], median, third quartile [Q3] and “maximum”). It can tell you about your outliers and what their values are. Boxplots can also tell you if your data is symmetrical, how tightly your data is grouped and if and how your data is skewed.

NOTE - Boxplot , Violinplot are very very important ,  we are gonna use this very frequent

image.png

```python
# Box plot
# axes level

sns.boxplot(
    data = tips,
    x = 'day' ,
    y = 'total_bill',

)
```

```python
# figure level

sns.catplot(
    data = tips,
    x ='day',
    y = 'total_bill',
    kind = 'box'
)
```

```python
# hue

sns.catplot(
    data = tips,
    x ='day',
    y = 'total_bill',
    kind = 'box',
    hue = 'sex'
)
```

```python
# single boxplot -> numerical col
sns.boxplot(data=tips,y='total_bill')
```

### Violinplot = (Boxplot + KDEplot)

```python
# axes level violin plot

sns.violinplot(
    data = tips,
    x = 'day',
    y = 'total_bill'
)


# the inner one which is the boxplot
# the outer one is kde plot
```

```python
# by figure level function

sns.catplot(
    kind = 'violin',
    data = tips,
    x = 'day',
    y = 'total_bill'
)
```

```python
# hue

sns.catplot(
    kind = 'violin',
    data = tips,
    x = 'day',
    y = 'total_bill',
    hue = 'sex'
)
```

```python
# split - works with hue

sns.catplot(
    kind = 'violin',
    data = tips,
    x = 'day',
    y = 'total_bill',
    hue = 'sex',
    split = True
)
```

Categorical Estimate Plot -> for central tendency


*   Barplot
*   Pointplot
*   Countplot




```python
# barplot
# A bar plot shows the mean of a numeric variable for each category, with optional error bars.


# axes level function
import numpy as np
sns.barplot(
    data = tips,
    x = 'sex',
    y = 'total_bill',
    hue = 'smoker',
    # estimator=np.min ,
    estimator = np.median
)

# estimator - any aggregate function name
```

```python
# figure level functon

sns.catplot(
    kind = 'bar' ,
    data = tips,
    x = 'sex',
    y = 'total_bill',
    hue = 'smoker',
    # estimator=np.min ,
    estimator = np.min
)
```

```python
sns.catplot(
    kind = 'bar' ,
    data = tips,
    x = 'sex',
    y = 'total_bill',
    hue = 'smoker',
    # estimator=np.min ,
    estimator = np.median ,
    ci = None
)

```

```python
# point plot
# it is brother of bar plot , where it takes the top part of all grapgh and make difference line between then
# it is not a simple line grapgh , it wroks on aggreagte funct
# simply understand ,  joint all top of bar plot , u will get this

sns.pointplot(
    data = tips,
    x = 'day',
    y = 'total_bill',
    errorbar = ('ci' , 0),
    estimator = np.min,
    hue = 'sex'
)
```

### **Point plot** - When there are multiple observations in each category, it also uses bootstrapping to compute a confidence interval around the estimate, which is plotted using error bars

```python
# countplot

sns.countplot(
    data = tips,
    x = 'sex',
    hue = 'day'
)

# returns the number of counts
```

**Count plot** - A special case for the bar plot is when you want to show the number of observations in each category rather than computing a statistic for a second variable. This is similar to a histogram over a categorical, rather than quantitative, variable

```python
# faceting using catplot

sns.catplot(
    data = tips ,
    x = 'sex',
    y = 'total_bill',
    kind = 'box' ,
    col = 'smoker',
    row = 'time',
    hue = 'sex'
)
```
