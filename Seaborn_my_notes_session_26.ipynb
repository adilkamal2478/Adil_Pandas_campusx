```python
# strip plot
# A strip plot is one of Seaborn’s categorical scatter plots
# axes level function

sns.stripplot(
    data = tips,
    x = 'day',
    y = 'total_bill',
)
```

```python
# jitter
# it remove unwanted jitter produced by seaborn in the graph

sns.stripplot(
    data = tips,
    x = 'day',
    y = 'total_bill',
    jitter = False
)
```

```python
# using catplot
# figure level function

sns.catplot(
    data = tips,
    x = 'day',
    y = 'total_bill',
    kind = 'strip'
)
```

```python
# changing jitter

sns.catplot(
    data = tips,
    x = 'day',
    y = 'total_bill',
    kind = 'strip',
    jitter = 0.2
)

# u can control the jitter as per ur requirement
```

```python
# hue

sns.catplot(
    data = tips,
    x = 'day',
    y = 'total_bill',
    kind = 'strip',
    hue = 'sex'
)
```

```python
# swarmplot - exactly same as strip plot
# Swarm plots automatically adjust the position of points so they don’t overlap.

# axes level
sns.swarmplot(data = tips , x = 'day' , y = 'total_bill')

# we can use this but for smaller data
```

```python
# figure level

sns.catplot( kind = 'swarm', data = tips , x = 'day' , y = 'total_bill')
```

```python
# hue

sns.catplot( kind = 'swarm', data = tips , x = 'day' , y = 'total_bill' , hue = 'sex')
```

### Boxplot

A boxplot is a standardized way of displaying the distribution of data based on a five number summary (“minimum”, first quartile [Q1], median, third quartile [Q3] and “maximum”). It can tell you about your outliers and what their values are. Boxplots can also tell you if your data is symmetrical, how tightly your data is grouped and if and how your data is skewed.

NOTE - Boxplot , Violinplot are very very important ,  we are gonna use this very frequent

image.png

```python
# Box plot
# axes level

sns.boxplot(
    data = tips,
    x = 'day' ,
    y = 'total_bill',

)
```

```python
# figure level

sns.catplot(
    data = tips,
    x ='day',
    y = 'total_bill',
    kind = 'box'
)
```

```python
# hue

sns.catplot(
    data = tips,
    x ='day',
    y = 'total_bill',
    kind = 'box',
    hue = 'sex'
)
```

```python
# single boxplot -> numerical col
sns.boxplot(data=tips,y='total_bill')
```
