```python
import seaborn as sns
```

```python
# import datasets
tips = sns.load_dataset('tips')
iris = sns.load_dataset('iris')
```

### Figure level function -> `catplot`

```python
sns.scatterplot(data = tips , x = 'total_bill' , y = 'tip')
```

### NOTE - if u want scatter plot on a categorical data , then one column shoulf be categorical and one should be numberic


> Add blockquote



```python
# strip plot
# A strip plot is one of Seaborn’s categorical scatter plots
# axes level function

sns.stripplot(
    data = tips,
    x = 'day',
    y = 'total_bill',
)
```

```python
# jitter
# it remove unwanted jitter produced by seaborn in the graph

sns.stripplot(
    data = tips,
    x = 'day',
    y = 'total_bill',
    jitter = False
)
```

```python
# using catplot
# figure level function

sns.catplot(
    data = tips,
    x = 'day',
    y = 'total_bill',
    kind = 'strip'
)
```

```python
# changing jitter

sns.catplot(
    data = tips,
    x = 'day',
    y = 'total_bill',
    kind = 'strip',
    jitter = 0.2
)

# u can control the jitter as per ur requirement
```

```python
# hue

sns.catplot(
    data = tips,
    x = 'day',
    y = 'total_bill',
    kind = 'strip',
    hue = 'sex'
)
```
