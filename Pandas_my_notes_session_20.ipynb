```python
# pd.concat

regs = pd.concat([nov,dec], ignore_index=True)
regs

# use concat in place of append everywhere
# nov index from 0 to 24 but during conactination , after nov the dec index also starts with 0 to 27
# so we use ignoreindex so that old index are replaced by new 0 to all index
```

```python
# multindex DataFrame

multi = pd.concat([nov,dec], keys=['Nov' , 'Dec'])
multi

# here we have made keys for each datasets like for nov we have key Nov
# here we have multi indexes , to acces using multindex we use loc()

multi.loc['Nov']      # all nov data
multi.loc['Dec']      # all dec data

multi.loc[('Dec' , 3)]    # 3rd index of dec
```

**Merge**

1. Left join

```python
# inner join
students
regs

students.merge(regs , how = 'inner' , on='student_id')
```

```python
# left join

courses
regs

courses.merge(regs , how='left' , on='course_id')
```

```python
# adding some student detail to understand all join better

temp_df = pd.DataFrame({
    'student_id':[26,27,28],
    'name':['Nitish','Ankit','Rahul'],
    'partner':[28,26,17]
})

students = pd.concat([students,temp_df],ignore_index=True)
```

```python
students.tail()
```

```python
# right join
print(students)
print(regs)
students.merge(regs , how='right' , on='student_id')
```

```python
# outer join (or full join)

students.merge(regs , how='outer' , on='student_id').tail(10)
```

```python
# 1. find total revenue generated
final = regs.merge(courses , how='inner' , on='course_id')
final['price'].sum()
```

```python
# 2. find month by month revenue

# nov_revenue = nov.merge(courses , how='inner' , on='course_id')['price'].sum()
# nov_revenue

temp_df = pd.concat([nov,dec],keys=['Nov','Dec']).reset_index()
temp_df.merge(courses,on='course_id').groupby('level_0')['price'].sum()
```

```python
# 3. Print the registration table
# cols -> name -> course -> price

regs
# courses
# students

students.merge(regs,on='student_id').merge(courses,on='course_id')[['name','course_name','price']]
```

```python
# 4. Plot bar chart for revenue/course

regs
courses

temp = regs.merge(courses,on='course_id')
temp = temp.groupby('course_name')['price'].sum()
temp.plot(kind='bar')
```

```python
# 5. find students who enrolled in both the months

nov
dec
# nov.merge(dec,how='inner' , on='student_id')['student_id'].drop_duplicates().sort_values()

common_stud = np.intersect1d(nov['student_id'],dec['student_id'])
```

```python
# 6. find course that got no enrollment
# courses['course_id']
# regs['course_id']

courses
regs

# temp = courses.merge(regs, how='left' , on='course_id')
# temp[temp['student_id'].isna()]

temp = np.setdiff1d(courses['course_id'] , regs['course_id'])
courses[courses['course_id'].isin(temp)]
```

```python
# 7. find students who did not enroll into any courses

students
regs

temp = students.merge(regs, how='left' , on='student_id')
temp[temp['course_id'].isna()]

students
regs
temp = np.setdiff1d(students['student_id'] , regs['student_id'])
students[students['student_id'].isin(temp)]
```

```python
# 8. Print student name -> partner name for all enrolled students
# self join

students.merge(students , how='inner',left_on='partner',right_on='student_id')[['name_x' , 'name_y']]
```

```python
# 9. find top 3 students who did most number enrollments

students
# regs.groupby('student_id')['course_id'].count().sort_values(ascending=False).head(3)

regs.merge(students , how = 'inner' , on = 'student_id').groupby(['student_id' , 'name'])['course_id'].count().sort_values(ascending=False).head(3)
```

```python
# 10. find top 3 students who spent most amount of money on courses

temp = regs.merge(students , how = 'inner' , on = 'student_id').merge(courses , how = 'inner' , on = 'course_id')
temp.groupby(['student_id' , 'name'])['price'].sum().sort_values(ascending=False).head(3)
```

```python
new_dl = delivery[delivery['batsman_runs'] == 6]
new_dl

all = matches.merge(new_dl , how = 'inner' , left_on='id' , right_on='match_id')

num_sixes = all.groupby('venue')['batsman_runs'].count()
num_sixes

num_matches = matches['venue'].value_counts()
num_matches

(num_sixes/num_matches).sort_values(ascending=False).head(10)
```

```python
# find orange cap holder of all the season

all = matches.merge(delivery , how = 'inner' , left_on='id' , right_on='match_id')
temp = all.groupby(['season','batsman'])['batsman_runs'].sum().reset_index().sort_values('batsman_runs' ,ascending=False)
temp.drop_duplicates(subset=['season'] , keep='first').sort_values('season')
```
