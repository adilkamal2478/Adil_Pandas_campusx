```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
iris = pd.read_csv('/content/drive/MyDrive/Matplotlib_CampusX/Session_24/iris.csv')
iris.head()
```

```python
iris['Species'] = iris['Species'].replace({'Iris-virginica' : 2 , 'Iris-setosa' : 0 , 'Iris-versicolor':1})

iris.sample(5)
```

```python
plt.scatter(iris['SepalLengthCm'] , iris['PetalLengthCm'] )

plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
```

```python
# color grouping
#    c=iris['Species'] , by this column grouping will happen
# NOTE - it only takes and int or float colun , it throw error on  string column
# so have replaced string with 0,1,2

plt.scatter(iris['SepalLengthCm'] , iris['PetalLengthCm'] , c=iris['Species'])

plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
```

```python
# colormap style
#  cmap='jet' -  its a style name
# some popular styles are - (viridis(default) ,  jet , inferno , plasma)

plt.scatter(iris['SepalLengthCm'] , iris['PetalLengthCm'] , c=iris['Species'] , cmap='jet')

plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
```

```python
# alpha - use to reduce or increase opacity or transparency
# alpha=0.7

plt.scatter(iris['SepalLengthCm'] , iris['PetalLengthCm'] , c=iris['Species'] , cmap='jet' , alpha = 0.4)

plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
```

```python
# colorbar()

plt.scatter(iris['SepalLengthCm'] , iris['PetalLengthCm'] , c=iris['Species'] , cmap='jet' , alpha = 0.9)

plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')

plt.colorbar()
```

### 3D Surface Plots

It is the one which is mostly used in Machine Learning


```python
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
x
y
```

```python
np.meshgrid(x,y)
```

```python
a = np.array([1, 2, 3, 4])
b = np.array([10, 20, 30,40,50])

my_x , my_y = np.meshgrid(a,b)
my_x
my_y

my_x*2 + my_y*2
```

```python
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)

xx , yy = np.meshgrid(x,y)

z = xx**2 + yy**2
print(z.shape)
z
```

```python
# z = x^2 + y^2

fig = plt.figure()
ax = plt.subplot(projection = '3d')

p = ax.plot_surface(xx,yy,z , cmap='viridis')

fig.colorbar(p)
```

```python
# z = sin(x) + cos(y)

z = np.sin(xx) + np.cos(yy)

fig = plt.figure(figsize=(12,8))

ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx,yy,z, cmap='viridis')
fig.colorbar(p)
```

```python
# z = log(x) + sin(y)

z = np.log(xx) + np.sin(yy)

fig = plt.figure(figsize=(12,8))

ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx,yy,z, cmap='viridis')
fig.colorbar(p)
```

```python
# z = sin(xx**2 + yy**2)

z = (np.sin(xx**2 + yy**2))

fig = plt.figure(figsize=(12,8))

ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx,yy,z, cmap='viridis')
fig.colorbar(p)
```

### Contour Plots

It represents `3d` grapgh into `2d`

```python
# contour graph
# the below graph is the 2d version of a 3d graph (z = x^2 + y^2)

z = xx**2 + yy**2
fig = plt.figure()
ax = plt.subplot()

p = ax.contour(xx,yy,z , cmap='viridis')

fig.colorbar(p)
```

```python
# contourf - means a countour graph which is filled with colors

z = xx**2 + yy**2
fig = plt.figure()
ax = plt.subplot()

p = ax.contourf(xx,yy,z , cmap='viridis')

fig.colorbar(p)
```

```python
z = np.sin(xx) + np.cos(yy)

fig = plt.figure(figsize=(12,8))
ax = plt.subplot()

p = ax.contourf(xx,yy,z)
plt.colorbar(p)
```
