```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
iris = pd.read_csv('/content/drive/MyDrive/Matplotlib_CampusX/Session_24/iris.csv')
iris.head()
```

```python
iris['Species'] = iris['Species'].replace({'Iris-virginica' : 2 , 'Iris-setosa' : 0 , 'Iris-versicolor':1})

iris.sample(5)
```

```python
plt.scatter(iris['SepalLengthCm'] , iris['PetalLengthCm'] )

plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
```

```python
# color grouping
#    c=iris['Species'] , by this column grouping will happen
# NOTE - it only takes and int or float colun , it throw error on  string column
# so have replaced string with 0,1,2

plt.scatter(iris['SepalLengthCm'] , iris['PetalLengthCm'] , c=iris['Species'])

plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
```

```python
# colormap style
#  cmap='jet' -  its a style name
# some popular styles are - (viridis(default) ,  jet , inferno , plasma)

plt.scatter(iris['SepalLengthCm'] , iris['PetalLengthCm'] , c=iris['Species'] , cmap='jet')

plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
```

```python
# alpha - use to reduce or increase opacity or transparency
# alpha=0.7

plt.scatter(iris['SepalLengthCm'] , iris['PetalLengthCm'] , c=iris['Species'] , cmap='jet' , alpha = 0.4)

plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
```

```python
# colorbar()

plt.scatter(iris['SepalLengthCm'] , iris['PetalLengthCm'] , c=iris['Species'] , cmap='jet' , alpha = 0.9)

plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')

plt.colorbar()
```

### Plot size

```python
plt.figure(figsize=(15,7))    # works on all types of graphs

plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap='jet',alpha=0.7)
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.colorbar()
```

```python
batters = pd.read_csv('/content/drive/MyDrive/Matplotlib_CampusX/Session_24/batter.csv')
batters.head()
```

```python
sample_df = batters.head(100).sample(25,random_state=5)
sample_df
```

```python
x = [1,2,3,4]
y = [5,6,7,8]

plt.scatter(x,y)
plt.text(1,5 , 'A' ,fontdict={'size':12,'color':'orange'})
plt.text(2,6 , 'B' ,fontdict={'size':26,'color':'purple'})
plt.text(3,7 , 'C' ,fontdict={'size':40,'color':'green'})
plt.text(4,8 , 'D' , fontdict={'size':25,'color':'brown'})
```

```python
plt.figure(figsize=(18,10))

plt.scatter(sample_df['avg'] , sample_df['strike_rate'] , s = sample_df['runs'])

for i in range(sample_df.shape[0]):
  plt.text(sample_df['avg'].values[i] , sample_df['strike_rate'].values[i] , sample_df['batter'].values[i])
```

### Horizontal and Vertical lines

plt.axhline()       - for horizontal line

plt.axvline()       - fro vertical line

```python
plt.figure(figsize=(18,10))

plt.scatter(sample_df['avg'] , sample_df['strike_rate'] , s = sample_df['runs'])

plt.axhline(130 , color='red')
plt.axvline(30, color='red')
plt.axhline(140, color='green')

for i in range(sample_df.shape[0]):
  plt.text(sample_df['avg'].values[i] , sample_df['strike_rate'].values[i] , sample_df['batter'].values[i])
```

### Subplots

```python
# normal way of plotting graph

plt.figure(figsize=(15,6))

plt.scatter(batters['avg'] , batters['strike_rate'])

plt.title('Something')
plt.xlabel('Avg')
plt.ylabel('Strike rate')
plt.show()
```

```python
# different way of plotting
# Object‑Oriented style
# this one is better approach

fig , ax = plt.subplots(figsize=(8,4))

ax.scatter(batters['avg'] , batters['strike_rate'] , color='red' , marker='+')

ax.set_title('Something')
ax.set_xlabel('Avg')
ax.set_ylabel('Strike rate')
fig.show()
```

```python
plt.subplots(nrows = 2 , ncols=2)
```

```python
fig , ax = plt.subplots(nrows = 2  ,ncols = 1 , sharex=True , figsize = (10,6))

ax[0].scatter(batters['avg'] , batters['strike_rate'] , color = 'red')
ax[1].scatter(batters['avg'] , batters['runs'])

ax[0].set_ylabel('Strike rate')
ax[1].set_ylabel('Runs')

# ax[0].set_xlabel('Avg')
ax[1].set_xlabel('Avg')

ax[0].set_title('Avg vs Strike rate')
ax[1].set_title('Avg vs Runs')
```

```python
# this is best way

fig , ax = plt.subplots(nrows = 2 , ncols = 2 ,figsize=(10,10))

ax[0,0].scatter(batters['avg'] , batters['strike_rate'] , color = 'red')
ax[0,1].scatter(batters['avg'] , batters['runs'] , color = 'blue')
ax[1,0].hist(batters['avg']  , color = 'green')
ax[1,1].hist( batters['runs'] , color = 'yellow')

# ax[0][0].set_ylabel('Strike rate')
# ax[0][0].set_ylabel('Strike rate')
```

```python
# Another way of making subplots but rarely used

fig = plt.figure()

ax1 = fig.add_subplot(2,2,1)
ax1.scatter(batters['avg'],batters['strike_rate'],color='red')

ax2 = fig.add_subplot(2,2,2)
ax2.hist(batters['runs'])

ax3 = fig.add_subplot(2,2,3)
ax3.hist(batters['avg'])
```
