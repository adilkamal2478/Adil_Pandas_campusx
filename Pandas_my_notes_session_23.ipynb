```python
# from a pandas dataframe
batsman = pd.read_csv('/content/drive/MyDrive/Matplotlib_CampusX/Session_23/sharma-kohli.csv')
batsman

plt.plot(batsman['index'] , batsman['RG Sharma'])
```

```python
# plotting multiple plots

plt.plot(batsman['index'] , batsman['RG Sharma'])
plt.plot(batsman['index'] , batsman['V Kohli'])
```

```python
# labels title

plt.plot(batsman['index'] , batsman['RG Sharma'])
plt.plot(batsman['index'] , batsman['V Kohli'])

plt.title('RoKo Career Comparison')

plt.xlabel('Season')
plt.ylabel('Runs scored')
```

```python
# colors(hex) and line(width and style) and marker(size)
plt.plot(batsman['index'],batsman['V Kohli'],color='red')
plt.plot(batsman['index'],batsman['RG Sharma'],color='blue') # u can use hex codes like #FC00D6

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
# line style
# linestyle (solid , dashed , dotted, dashdot)
# linewidth - it manages the width of line
# marker - every top part of graph (D-diamond, o ,'+' , '.' , '>') generally o is used
# markersize - size of marker

plt.plot(batsman['index'],batsman['V Kohli'],color='red' , linestyle = 'solid' , linewidth=5, marker='o' , markersize = 10)
plt.plot(batsman['index'],batsman['RG Sharma'],color='blue' , linestyle = 'dashed' ,linewidth=2)

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
# legend -> location

plt.plot(batsman['index'],batsman['V Kohli'],color='red' , linestyle = 'solid' , linewidth=5, marker='o' , markersize = 10 , label ='Virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color='blue' , linestyle = 'dashed' ,linewidth=2 , label='Rohit')

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')

plt.legend()
plt.legend(loc = 'upper right')

```

```python
# limiting axes
price = [48000,54000,57000,49000,47000,45000,4500000]
year = [2015,2016,2017,2018,2019,2020,2021]

plt.plot(year,price)

plt.ylim(0,75000)
plt.xlim(2017,2019)

# it limits the x and y axis according to our use
```

```python
# grid
plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F',linestyle='solid',linewidth=3,marker='D',markersize=10)
plt.plot(batsman['index'],batsman['RG Sharma'],color='#FC00D6',linestyle='dashdot',linewidth=2,marker='o')

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')

plt.grid()    # maked grid in graph
```

```python
# show
plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F',linestyle='solid',linewidth=3,marker='D',markersize=10)
plt.plot(batsman['index'],batsman['RG Sharma'],color='#FC00D6',linestyle='dashdot',linewidth=2,marker='o')

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')

plt.grid()

plt.show()  #in most of the editors we use show() to see the graph, without this graph will not show
```

```python
# plt.scatter on pandas data
df = pd.read_csv('/content/drive/MyDrive/Matplotlib_CampusX/Session_23/batter.csv')
df = df.head(50)
df
```

```python
plt.scatter(df['avg'] , df['strike_rate'] , color='red',marker='+')

plt.title('Avg and SR analysis of Top 50 Batsman')
plt.xlabel('Avg')
plt.ylabel('Strike rate')
```

```python
# size
tips = sns.load_dataset('tips')
tips
```

```python
# slower

plt.scatter(tips['total_bill'],tips['tip'] , s= tips['size']*20)

# size of  single dot represents the amount of person come
```

```python
# scatterplot using plt.plot
# faster
plt.plot(tips['total_bill'],tips['tip'] , 'o')
```
