```python
# import titanic

df = pd.read_csv('/content/drive/MyDrive/Pandas_CampusX/Session_22/titanic.csv')
df['Name']
```

```python
# Common Functions
# lower/upper/capitalize/title
# len
# strip
```

```python
# lower/upper/capitalize/title
df['Name'].str.lower()              # all small
df['Name'].str.upper()              # all capital
df['Name'].str.capitalize()         # first letter of a line becomes capital
df['Name'].str.title()         # first letter of ech word becomes capital

```

```python
# len
df['Name'].str.len()
```

```python
# strip (removes all leading and trailing spaces)

"         adil               ".strip()

df['Name'].str.strip()
```

```python
# split -> get
df['Name'].str.split(',').str.get(0)
df['lastname'] = df['Name'].str.split(',').str.get(0)
df.head()
```

```python
df['Name'].str.split(',').str.get(1).str.strip().str.split(' ', n =1 , expand=True)
df[['title' , 'firstname']] = df['Name'].str.split(',').str.get(1).str.strip().str.split(' ', n =1 , expand=True)
df.head()
```

```python
df['title'].value_counts()
```

```python
# replace


df['title'] = df['title'].str.replace('Ms.' , 'Miss.')        # all Ms. replaced with Miss.
df['title'] = df['title'].str.replace('Mlle.' , 'Miss.')        # all Ms. replaced with Miss.
df['title'].value_counts()
```

```python
# filtering
# startswith/endswith
df[df['firstname'].str.startswith('A')]
df[df['firstname'].str.endswith('m')]
# isdigit/isalpha...
df[df['firstname'].str.isdigit()]
```

```python
# Advance filtering

# applying regex
# contains

# search john -> both case
df[df['firstname'].str.contains('john',case=False)]

# find lastnames with start and end char vowel
df[df['lastname'].str.contains('^[aeiouAEIOU].+[aeiouAEIOU]$')]
```

```python
# slicing
df['Name'].str[:4]
df['Name'].str[::2]
df['Name'].str[::-1]
```
