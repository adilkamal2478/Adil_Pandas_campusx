```python
# agg method
# passing dict
genres.agg(
    {
        'Runtime':'mean',
        'IMDB_Rating':'mean',
        'No_of_Votes':'sum',
        'Gross':'sum',
        'Metascore':'min'
    }
)

# applying different aggregate functions on different columns at a time
```

```python
# passing list (this will apply all given aggregate function on each column)


genres[['Runtime','IMDB_Rating','No_of_Votes','Gross','Metascore']].agg(['min','max','sum','mean'])
# movies

# NOTE - we have to pass column names in a list so that the string columns doesnt produce any error
# pass mostly numeric columns
```

```python
# passing both dict and list
# upper both are combined together

genres.agg(
    {
        'Runtime':'mean',
        'IMDB_Rating':'mean',
        'No_of_Votes':['min','max','sum'],
        'Gross':'sum',
        'Metascore':'min'
    }
)

# here No_of_Votes columns will be divided into 3 columns 'min','max','sum'
```

```python
# looping on groups
rows = []

for group,data in genres:
  rows.append(data[data['IMDB_Rating'] == data['IMDB_Rating'].max()])

df_result = pd.concat(rows, ignore_index=True)
df_result
```

```python
# split (apply) combine
# apply -> builtin function

genres.apply(min)
```

```python
# find number of movies starting with A for each group
def str_A(x):
  return x['Series_Title'].str.startswith('A').sum()

genres.apply(str_A)
```

```python
# find ranking of each movie in the group according to IMDB score

def rank_movies(group):
  group['genres_rank'] = group['IMDB_Rating'].rank(ascending=False)
  return group


genres.apply(rank_movies)
```

```python
# Find normalized rating group wise

def normalized(group):
  X_min = group['IMDB_Rating'].min()
  X_max = group['IMDB_Rating'].max()

  group['norm_rating'] = (group['IMDB_Rating'] - X_min) / (X_max - X_min)
  return group

genres.apply(normalized)
```

```python
ipl = pd.read_csv('/content/drive/MyDrive/Pandas_CampusX/Session_19/deliveries.csv')
ipl.head(20)
```

```python
ipl.shape
```

```python
# Find the top 10 batsman in terms of runs

ipl.groupby('batsman')['batsman_runs'].sum().sort_values(ascending=False).head(10)
```

```python
# Find the batsman with max no of sixes
six = ipl[ipl['batsman_runs'] == 6]
six.groupby('batsman')['batsman_runs'].count().sort_values(ascending=False).head(3)
```

```python
# Find the batsman with most number of 4's and 6's in last 5 overs

last_five = ipl[ipl['over'] > 15 & ipl['batsman_runs'].isin([4,6])]

last_five.groupby('batsman')['batsman_runs'].count().sort_values(ascending=False)
```

```python
# Find v kohli's record against all teams

kohli = ipl[ipl['batsman'] == 'V Kohli']
kohli.groupby(['batsman' , 'bowling_team'])['batsman_runs'].sum().sort_values(ascending=False)
```

```python
# create a function that can return the highest score of any batsman

def highest(name):
  filter = ipl[ipl['batsman'] == name ]
  runs = filter.groupby('match_id')['batsman_runs'].sum().sort_values(ascending=False).head(1).values[0]
  print(name ,'highest run is:',runs)

highest('V Kohli')
```
