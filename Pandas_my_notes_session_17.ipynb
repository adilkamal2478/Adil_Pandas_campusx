### Selecting rows from a DataFrame

- **iloc** - searches using index positions
- **loc** - searches using index labels

```python
# single row
movies.iloc[0]    # 1st movie
```

```python
# multiple row

movies.iloc[0:5]    # 1st 5 movie
movies.iloc[5:16:2]    # 5th 7th ... 15th
```

```python
# fancy indexing
movies.iloc[[0,1]]    # 1st 2nd movie
movies.iloc[[0,1,6]]    # 1st 2nd 7th movie
```
