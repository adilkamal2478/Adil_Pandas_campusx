### **`Problem-1:`** Draw a line plot of which, the x-axis is the "Year" and the y-axis is sum of "PM2.5" of two countries Iran and China.

```python
# code here
df = pd.read_csv('https://tinyurl.com/2fe6vz4u')
df.head()
```

```python
from matplotlib.lines import lineStyles
temp1 = df[df['Country'] == 'Iran']
temp2 = df[df['Country'] == 'China']
temp1 = temp1.groupby('Year')['PM2.5'].sum().reset_index()
temp2 = temp2.groupby('Year')['PM2.5'].sum().reset_index()

plt.plot(temp1['Year'] , temp1['PM2.5'] , linestyle = 'dashed' , linewidth = 3 , label = 'Iran')
plt.plot(temp2['Year'] , temp2['PM2.5'] ,  linewidth = 3 , label = 'China')

plt.xticks(df['Year'])
plt.xlabel('Year')
plt.ylabel('PM2.5')
plt.title('PM2.5 Over the years')
plt.legend()
plt.grid()
plt.show()
```

### **`Problem-2:`** Draw a histogram of the  column "PM10" of which the y-axis represents the probability (see the documentation how to draw the probability).

```python
df.head(3)
```

```python
plt.hist(df['PM10'] ,bins = 50  , density = True)
plt.xlabel('Bins')
plt.ylabel('Probability')
plt.title('Histogram of PM10')
plt.grid()
plt.show()
```

### **`Problem-3:`** Draw a scatter plot where x-axis represents "PM2.5" and y-axis represents "PM10" for two countries Poland and Chile.

```python
df.head()
poland = df[df['Country'] == 'Poland']
chile = df[df['Country'] == 'Chile']
```

```python
plt.scatter(poland['PM2.5'] , poland['PM10'] , marker = 'D' , color = 'red' ,label='Poland' )
plt.scatter(chile['PM2.5'] , chile['PM10'] , marker = '+' , color = 'black'  ,label='Chile')

plt.xlabel('PM2.5')
plt.ylabel('PM10')
plt.title('PM2.5 Vs PM10 for Chile and Poland')
plt.legend()
plt.grid()
plt.show()
```

### **`Problem-4:`** Draw a pie chart of top 5 most frequent countries.

```python
df
```

```python
temp = df.groupby('Country')['Position'].count().sort_values(ascending=False).head(5).reset_index()
temp
```

```python
plt.pie(temp['Position'] , labels= temp['Country'] , autopct = '%0.1f%%')

plt.show()
```

### **`Problem-5:`** Draw a bar chart which represents the counts of top 5 most frequent countries.
